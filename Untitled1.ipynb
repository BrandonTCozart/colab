```python
from google.colab import files
import pandas as pd
import io

uploaded = files.upload()

filename = next(iter(uploaded))

df = pd.read_csv(
    io.BytesIO(uploaded[filename]),
    header=None,
    names=["Date", "Open", "High", "Low", "Close", "Volume"]
)

df.head()
```

```python
# Convert date column
df["Date"] = pd.to_datetime(df["Date"], utc=True)

# Convert UTC -> Eastern Time
df["Date_ET"] = df["Date"].dt.tz_convert("America/New_York")

# Create separate date/time columns
df["Calendar_Date"] = df["Date_ET"].dt.date
df["Time"] = df["Date_ET"].dt.time

# Get 6:00 PM rows
open_rows = df[
    (df["Date_ET"].dt.hour == 18) &
    (df["Date_ET"].dt.minute == 0)
].copy()

# The 6 PM session belongs to the NEXT trading day
open_rows["Trading_Day"] = (
    open_rows["Date_ET"].dt.normalize() + pd.Timedelta(days=1)
).dt.date

open_rows = open_rows[
    ["Trading_Day", "Open"]
].rename(columns={
    "Open": "Open_6PM"
})

# Get 4:00 PM rows
close_rows = df[
    (df["Date_ET"].dt.hour == 16) &
    (df["Date_ET"].dt.minute == 0)
].copy()

close_rows["Trading_Day"] = close_rows["Date_ET"].dt.date

close_rows = close_rows[
    ["Trading_Day", "Close"]
].rename(columns={
    "Close": "Close_4PM"
})

# Combine the open and close
daily = pd.merge(
    open_rows,
    close_rows,
    on="Trading_Day",
    how="inner"
)

# Calculate movement
daily["Move"] = daily["Close_4PM"] - daily["Open_6PM"]

# Optional percentage move
daily["Move_Percent"] = (
    daily["Move"] / daily["Open_6PM"] * 100
)

daily["Trading_Day"] = pd.to_datetime(daily["Trading_Day"])

daily.head()
```

```python
daily["Month"] = daily["Trading_Day"].dt.to_period("M")

results = []

for month, group in daily.groupby("Month"):

    best = group.loc[group["Move"].idxmax()]
    worst = group.loc[group["Move"].idxmin()]

    results.append({
        "Month": str(month),

        "Best_Day": best["Trading_Day"].date(),
        "Best_Open_6PM": best["Open_6PM"],
        "Best_Close_4PM": best["Close_4PM"],
        "Best_Move": best["Move"],
        "Best_Move_%": best["Move_Percent"],

        "Worst_Day": worst["Trading_Day"].date(),
        "Worst_Open_6PM": worst["Open_6PM"],
        "Worst_Close_4PM": worst["Close_4PM"],
        "Worst_Move": worst["Move"],
        "Worst_Move_%": worst["Move_Percent"]
    })

results_df = pd.DataFrame(results)

results_df
```
